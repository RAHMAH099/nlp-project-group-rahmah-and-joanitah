## A3 · Build a Gradio demo

Gradio turns a Python function into a web app. First test it here in Colab.

```
import gradio as gr

def predict(text):
    if not text.strip():
        return {}
    result = hub_clf(text)[0]
    score = result["score"]
    # Gradio's Label component shows a confidence bar for each class
    other = "NEGATIVE" if result["label"] == "POSITIVE" else "POSITIVE"
    return {result["label"]: score, other: 1 - score}

demo = gr.Interface(
    fn=predict,
    inputs=gr.Textbox(lines=3, placeholder="Type a sentence..."),
    outputs=gr.Label(num_top_classes=2),
    title="Sentiment Analyser",
    description="Type a sentence and see if it is positive or negative.",
    examples=["The food was amazing!", "Worst service ever.", "It was okay, nothing special."],
)
demo.launch()
```

## A5 · Turn your model into an API with FastAPI

A demo is for **people**. An API is for **other programs** (a mobile app, a website, a WhatsApp bot). They send JSON, your API sends JSON back.

We'll run it here in Colab so you can test it. On your laptop you run the same file with `uvicorn main:app --reload` and open `http://127.0.0.1:8000/docs`.

```
%%writefile main.py
from fastapi import FastAPI, HTTPException
from pydantic import BaseModel
from transformers import pipeline

MODEL_ID = "RahmahwithCode/rahmahandjoanitah-sentiment-model"   # TODO: your model id

app = FastAPI(title="Sentiment API")

# Load the model ONCE when the app starts, not on every request (that would be very slow)
clf = pipeline("text-classification", model=MODEL_ID)

class TextIn(BaseModel):
    text: str

class PredictionOut(BaseModel):
    label: str
    score: float

@app.get("/health")
def health():
    return {"status": "ok"}

@app.post("/predict", response_model=PredictionOut)
def predict(item: TextIn):
    if not item.text.strip():
        raise HTTPException(status_code=400, detail="Text cannot be empty")
    result = clf(item.text)[0]
    return PredictionOut(label=result["label"], score=round(result["score"], 4))
```

```
from transformers import pipeline

pipe = pipeline(
    "text-generation",
    model="Qwen/Qwen2.5-1.5B-Instruct",
    device_map="auto",
    torch_dtype="auto",
)

def ask_llm(messages, max_tokens=400):
    out = pipe(messages, max_new_tokens=max_tokens, do_sample=False)
    return out[0]["generated_text"][-1]["content"]

print(ask_llm([
    {"role": "system", "content": "You are a helpful assistant. Answer in one sentence."},
    {"role": "user", "content": "What is natural language processing?"},
]))
```

## B2 · RAG: answer questions from your PDF

LLMs don't know about *your* documents and may invent answers. **Retrieval-Augmented Generation (RAG)** fixes this:

1. Split the document into small chunks
2. Turn each chunk into an embedding (like the BERT embeddings in the tutorial)
3. When a question comes in, find the most similar chunks
4. Give those chunks to the LLM and tell it to answer **only** from them

### Upload your group's PDF
Each group uses a **different** document (course handbook, hostel rules, a health guide, a mobile-money FAQ, a farming guide, ...). Click the 📁 folder icon on the left and upload it, or run the cell below.

```
from google.colab import files
uploaded = files.upload()
PDF_PATH = list(uploaded.keys())[0]
print("Using:", PDF_PATH)
```

```
from pypdf import PdfReader

reader = PdfReader(PDF_PATH)
pages = [(i + 1, page.extract_text() or "") for i, page in enumerate(reader.pages)]
print(f"{len(pages)} pages. First 300 characters of page 1:\n")
print(pages[0][1][:300])
```

### Step 1 · Split into chunks (keeping the page number)

```
def chunk_pages(pages, chunk_size=500, overlap=100):
    chunks = []
    for page_num, text in pages:
        text = " ".join(text.split())          # clean up extra spaces and line breaks
        start = 0
        while start < len(text):
            piece = text[start:start + chunk_size]
            if len(piece) > 50:                # skip tiny leftovers
                chunks.append({"page": page_num, "text": piece})
            start += chunk_size - overlap      # overlap so sentences aren't cut in half
    return chunks

chunks = chunk_pages(pages)
print(len(chunks), "chunks")
print(chunks[0])
```

### Step 2 · Embed the chunks

`sentence-transformers` models are trained so that sentences with similar meaning get similar vectors, which makes them better for search than the mean-pooled BERT from the tutorial.

```
from sentence_transformers import SentenceTransformer
import numpy as np

embedder = SentenceTransformer("all-MiniLM-L6-v2")
chunk_vectors = embedder.encode([c["text"] for c in chunks], normalize_embeddings=True)
print(chunk_vectors.shape)   # (number of chunks, 384)
```

### Step 3 · Retrieve the most relevant chunks for a question

```
def retrieve(question, k=3):
    q_vec = embedder.encode([question], normalize_embeddings=True)[0]
    scores = chunk_vectors @ q_vec            # cosine similarity (vectors are normalised)
    top = np.argsort(scores)[::-1][:k]
    return [{**chunks[i], "score": float(scores[i])} for i in top]

question = "How will exercise improve my health?"   # TODO: a question about YOUR document
for hit in retrieve(question):
    print(f"page {hit['page']}  score={hit['score']:.2f}  {hit['text'][:120]}...")
```

### Step 4 · Let the LLM answer from those chunks only

```
SYSTEM_PROMPT = (
    "You answer questions using ONLY the context provided. "
    "Cite page numbers like (page 3). "
    "If the answer is not in the context, say: I don't know, that is not in the document."
)

def rag_answer(question, k=3):
    hits = retrieve(question, k)
    context = "\n\n".join(f"[page {h['page']}] {h['text']}" for h in hits)
    return ask_llm([
        {"role": "system", "content": SYSTEM_PROMPT},
        {"role": "user", "content": f"Context:\n{context}\n\nQuestion: {question}"},
    ])

print(rag_answer(question))
```

## B3 · Turn it into a chatbot

`gr.ChatInterface` gives you a chat window. We pass the **conversation history** to the LLM so it remembers earlier questions (e.g. *"What about the second one?"*).

```
!pip install --upgrade gradio
```

```

SYSTEM_PROMPT = """
You are Health Guide Assistant, a polite and helpful document-based chatbot.

1. Answer questions about the uploaded document using ONLY the retrieved context.
2. Use conversation history to understand follow-up questions.
3. If a question is unrelated to the document, politely explain your purpose and redirect the user.
4. Respond politely to greetings.
5. Explain your identity when asked.
6. Remain professional when users are rude.
7. Never invent information.
8. ALWAYS include the page number when answering questions using document information.
9. Use this citation format: (Page 7).
10. If information is found on multiple pages, cite all relevant page numbers.
11. If information is unavailable, clearly say: "I don't know, that is not in the document."

Your main purpose is to help users understand the uploaded health document.
"""
```

```
import gradio as gr

NOT_FOUND = "I don't know. That is not in the document."

def chat(message, history):

    # Initialize history
    history = history or []

    # Retrieve relevant information from the document
    hits = retrieve(message)

    # If nothing was retrieved, the answer is not in the document
    if not hits:
        return NOT_FOUND

    # Combine retrieved document sections
    context = "\n\n".join(
        f"[Page {h['page']}] {h['text']}"
        for h in hits
    )

    # Prepare system instructions
    messages = [{
        "role": "system",
        "content": (
            SYSTEM_PROMPT
            + "\nUse conversation history to understand follow-up questions."
            + "\nAnswer ONLY using the retrieved document context."
            + "\nIf the answer is in the context, give it and end with the page numbers"
            + " you used, like this: (Page 3) or (Pages 3, 5)."
            + " Only cite pages that actually support your answer."
            + f"\nIf the answer is NOT in the context, reply exactly: \"{NOT_FOUND}\""
            + " Do not guess, do not use outside knowledge, and do not cite any page."
        )
    }]

    # Keep the last 10 messages
    # Handle Gradio message-format history
    for item in history[-10:]:
        if isinstance(item, dict) and item.get("role") in ["user", "assistant"]:
            content = item.get("content", "")

            if isinstance(content, str):
                messages.append({
                    "role": item["role"],
                    "content": content
                })

    # Add the current question and document context
    messages.append({
        "role": "user",
        "content": (
            f"Document context:\n{context}\n\n"
            f"Question: {message}"
        )
    })

    # Generate the answer
    return ask_llm(messages)


# Create a fresh Gradio chatbot
bot = gr.ChatInterface(
    fn=chat,
    title="Health Guide Assistant",
    description="Ask questions about the uploaded document.",
    examples=[
        "What is this document about?",
        "What does it say about healthy eating?",
        "Can you explain that in simpler terms?"
    ]
)

bot.launch()
```

1. Added conversational memory: The history parameter was introduced to allow the chatbot to remember previous messages and responses during an active conversation.

2. Added conversation history management: The chatbot was modified to retain the last 10 messages to provide context for follow-up questions.

3. Improved document retrieval: The existing retrieve() function was integrated into the chatbot so that relevant document sections are retrieved whenever a user asks a question.

4. Combined document context and conversation history: Retrieved document information and previous conversation messages are passed together to the language model through ask_llm().

5. Updated the system prompt: The original instructions were expanded to define the chatbot's identity, behaviour and responsibilities.

6. Added off-topic question handling: The chatbot was instructed to politely redirect questions unrelated to the uploaded document.

7. Improved response behaviour: Instructions were added to handle greetings, respond professionally to rude messages, understand follow-up questions and avoid inventing information.

8. Updated the Gradio interface: The chatbot interface was adjusted for compatibility with Gradio 6.31.0 by removing the unsupported type="messages" argument.

9. Added example questions: Sample questions were included to help users understand how to interact with the chatbot.

## B4 · Deploy the chatbot to Spaces

The Space needs your PDF, the app code, and your token stored as a **secret** (never inside the code).

```

%%writefile chatbot_app.py
import os
import numpy as np
import gradio as gr
from pypdf import PdfReader
from sentence_transformers import SentenceTransformer
from huggingface_hub import InferenceClient

PDF_PATH = "document.pdf"
LLM_MODEL = "meta-llama/Llama-3.1-8B-Instruct"

SYSTEM_PROMPT = """
You are Health Guide Assistant, a polite and helpful document-based chatbot.

1. Answer questions about the uploaded document using ONLY the retrieved context.
2. Use conversation history to understand follow-up questions.
3. If a question is unrelated to the document, politely explain your purpose and redirect the user.
4. Respond politely to greetings.
5. Explain your identity when asked.
6. Remain professional when users are rude.
7. Never invent information.
8. Cite page numbers when providing document information.
9. If information is unavailable, clearly say so.
10. Returns the page number when answering questions using document information.

Your main purpose is to help users understand the uploaded health document.
"""

llm = InferenceClient(api_key=os.environ["HF_TOKEN"])   # read from the Space secret
embedder = SentenceTransformer("all-MiniLM-L6-v2")

pages = [(i + 1, p.extract_text() or "") for i, p in enumerate(PdfReader(PDF_PATH).pages)]
chunks = []
for page_num, text in pages:
    text = " ".join(text.split())
    for start in range(0, len(text), 400):
        piece = text[start:start + 500]
        if len(piece) > 50:
            chunks.append({"page": page_num, "text": piece})
vectors = embedder.encode([c["text"] for c in chunks], normalize_embeddings=True)

def retrieve(q, k=3):
    scores = vectors @ embedder.encode([q], normalize_embeddings=True)[0]
    return [chunks[i] for i in np.argsort(scores)[::-1][:k]]

def chat(message, history):
    context = "\n\n".join(f"[page {h['page']}] {h['text']}" for h in retrieve(message))
    messages = [{"role": "system", "content": SYSTEM_PROMPT}] + history[-10:]
    messages.append({"role": "user", "content": f"Context:\n{context}\n\nQuestion: {message}"})
    r = llm.chat.completions.create(model=LLM_MODEL, messages=messages, max_tokens=400, temperature=0.2)
    return r.choices[0].message.content

gr.ChatInterface(fn=chat, type="messages", title="Health Guide Assistant").launch()
```
